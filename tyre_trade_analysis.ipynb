## Merging Export and Import Datasets and Calculating Trade Balance

Match `e_comp` and `m_comp` by `Country` using an inner merge, retaining only partners present in both complete-data tables.

Count the partners excluded from each table during matching. The matched dataset will be used to calculate annual trade balances as exports minus imports, measured in A$ thousands.

```python
# Using inner merge, match countries present in both complete datasets
trade_matched = e_comp.merge(
    m_comp,
    on="Country",
    how="inner",
    suffixes=("_EX", "_IM")
) 
# Check the total number of countries that are present in both datasets
print(len(trade_matched))
trade_matched.head(8)
```

```python
# To rearrange the column to 2006_EX, 2006_IM
column_order = ["Country"]

for year in range(2006,2026):
    column_order.append(f"{year}_EX")
    column_order.append(f"{year}_IM")

trade_matched = trade_matched[column_order]

trade_matched.head(8)
```

```python
# To find the number of countries from import and export datasets that are not matched
ex_unmatched = e_comp.loc[
    ~e_comp["Country"].isin(trade_matched["Country"])
].copy()

im_unmatched = m_comp.loc[
    ~m_comp["Country"].isin(trade_matched["Country"])
].copy()

print("Export countries excluded:", len(ex_unmatched))
print(ex_unmatched["Country"].tolist())

print("Import countries excluded:", len(im_unmatched))
print(im_unmatched["Country"].tolist())

print("Total Numbers of Countries matched:", len(trade_matched))
```

```python
# To calculate the Trade Balance across each Year
for year in range(2006,2026): # Loop across years to calculate the trade
    trade_matched[f"{year}_TB"] = (
        trade_matched[f"{year}_EX"]-trade_matched[f"{year}_IM"]
    )

# To group the columns by years
column_order = ["Country"]

for year in range(2006,2026):
    column_order.append(f"{year}_EX")
    column_order.append(f"{year}_IM")
    column_order.append(f"{year}_TB")
trade_matched = trade_matched[column_order]
trade_matched.head()
```

## Trade Balance Analysis

Among the 22 matched trading partners, with which partners does Australia have a tyre-trade deficit, surplus or balance, and how have these balances changed over 2006–2025?

First, classify each partner’s 2025 trade balance as a surplus, deficit or balance. Then count the number of surplus, deficit and balanced years for each partner across 2006–2025 to identify persistent patterns.

- **Surplus:** exports exceed imports (`TB > 0`).
- **Deficit:** imports exceed exports (`TB < 0`).
- **Balance:** recorded exports equal imports (`TB = 0`).

All trade values are measured in A$ thousands. Findings cover only the 22 partners with complete data, not all Australian tyre trade.

### To find 2025 Trade Balance

```python
# To find the 2025 trade balance and classify the trade balance
bal_2025 = trade_matched[["Country","2025_TB"]].copy()

bal_2025.loc[bal_2025["2025_TB"] > 0, "Status"] = "Surplus"
bal_2025.loc[bal_2025["2025_TB"] < 0, "Status"] = "Deficit"
bal_2025.loc[bal_2025["2025_TB"] == 0, "Status"] = "Balanced"

# Show largest deficit first and largest surpluses last
bal_2025 = bal_2025.sort_values("2025_TB").reset_index().drop(columns="index")
bal_2025
```

#### Observations: 2025 Trade Balances

- Australia recorded tyre-trade deficits with **20 of the 22** matched partners. Only **New Zealand and the United Arab Emirates** recorded surpluses.

- China had the largest deficit at approximately *AUD 1.187 billion*, followed by **Japan** at *AUD 986.340 million* and **the United States** at *AUD 612.046 million*.

- **New Zealand** had the largest surplus at *AUD 21.065 million*, compared with *AUD 1.913 million* for the **United Arab Emirates**.

- **Hong Kong** had the smallest deficit at *AUD 253,000*. No partner had an exactly balanced trade position.

These findings cover only the 22 partners with complete data, not all Australian tyre trade.

### To count how many years each country had a surplus, deficit or balance across 2006-2025

```python
# Select 20 trade-balance columns
tb_columns = [f"{year}_TB" for year in range(2006,2026)]
bal = trade_matched[tb_columns]

# Count each outcome for every country
bal_counts = trade_matched[["Country"]].copy()

bal_counts["Surplus_years"] = (bal > 0).sum(axis=1)
bal_counts["Deficit_years"] = (bal < 0).sum(axis=1)
bal_counts["Balanced_years"] = (bal == 0).sum(axis=1)

# Show countries with the most surplus years first
bal_counts = bal_counts.sort_values(by="Surplus_years", ascending=False).reset_index().drop(columns="index")

bal_counts
```

#### Observations: Trade Balance Patterns, 2006–2025

- **Australia** recorded tyre-trade deficits with *16 of the 22* matched partners in every year from *2006 to 2025*.

- **New Zealand** was the only partner with more surplus years than deficit years, recording 19 surplus years and one deficit year.

- **Hong Kong**, **South Africa** and **the United Arab Emirates** each recorded nine surplus years and 11 deficit years. **Singapore** recorded seven surplus years, while the **Netherlands** recorded five.

- No partner recorded an exactly balanced trade position in any year.

These counts describe how frequently surpluses and deficits occurred, not their monetary size or when changes occurred. Findings apply only to the 22 partners with complete data.

## Interactive Chart: Australia’s Tyre Trade Balances by Partner, 2006–2025

```python
# Turn the yearly trade-balance columns into rows
trade_long = trade_matched.melt(
    id_vars="Country",
    value_vars=tb_columns,
    var_name="Year",
    value_name="Trade_balance"
)

# Convert labels such as "2006_TB" into integer 2006
trade_long["Year"] = (
    trade_long["Year"].str.replace("_TB","",regex=False).astype(int)
)

trade_long.head()
```

```python
# To plot for each country: plot trade balance on year
fig = px.line(
    trade_long.sort_values("Year"),
    x="Year",
    y="Trade_balance",
    color="Country",
    title="Australia’s Tyre Trade Balances by Partner, 2006–2025",
    labels={"Trade_balance": "Trade balance (A$ thousands)"}
)

# Add a reference line separating surplus and deficit
fig.add_hline(y=0, line_dash="dash", line_color="grey")

fig.show()
```

### Observations: Trade Balances Over Time

Australia’s tyre-trade deficits with China, Japan and the United States widened overall between 2006 and 2025, although their balances fluctuated during the period.

China’s deficit grew substantially, reaching approximately AUD 1.187 billion in 2025.

Many other partners appear close to zero because their trade balances are much smaller in magnitude. This does not mean their balances remained constant; viewing selected countries separately makes their changes easier to see.

These findings describe the 22 matched partners and do not establish the causes of the changes.

```python
# To make the graph more interactive, allow the user to enter a country name and the graph display that particular country's chart
# Creates 3 interface components: A text box, A button and an output area
country_input = widgets.Text(
    value="Canada",
    placeholder="Canada, Japan, New Zealand",
    description="Countries:",
    layout=widgets.Layout(width="600px")
)
plot_button = widgets.Button(description="Plot")
chart_output = widgets.Output()
```

```python
def update_chart(button):
    # Read the country names entered in the box
    countries = [
        name.strip()
        for name in country_input.value.split(",")
        if name.strip()
    ]

    with chart_output:
        clear_output(wait=True)

        # Check the names
        available = trade_long["Country"].unique()
        unknown = [name for name in countries if name not in available]

        if not countries:
            print("Enter at least one country.")
            return

        if unknown:
            print("Country names not found:", ", ".join(unknown))
            return

        # Select those countries and plot their balances
        selected = trade_long[
            trade_long["Country"].isin(countries)
        ].sort_values("Year")

        fig = px.line(
            selected,
            x="Year",
            y="Trade_balance",
            color="Country",
            markers=True,
            title="Australia’s Tyre Trade Balances by Partner",
            labels={"Trade_balance": "Trade balance (A$ thousands)"}
        )

        fig.add_hline(y=0, line_dash="dash", line_color="grey")
        fig.show(renderer="vscode")


plot_button.on_click(update_chart)
```

```python
plot_button.on_click(update_chart)

display(country_input, plot_button, chart_output)
```

### Changes in Trade Balances Between 2006 and 2025

Compare each partner’s trade balance in 2006 and 2025 to identify the largest increases and decreases.

Change in trade balance = 2025 trade balance − 2006 trade balance.

A positive change indicates a larger surplus or smaller deficit, while a negative change indicates a smaller surplus or larger deficit. A partner may also switch between surplus and deficit.

Values are in A$ thousands. This comparison covers the 22 matched partners and measures the change between the two endpoints, rather than fluctuations throughout the period.

```python
bal_comp = trade_matched[["Country","2006_TB","2025_TB"]].copy()
bal_comp["Change"]= bal_comp["2025_TB"] - bal_comp["2006_TB"] 
bal_comp.sort_values("Change", ascending=True).reset_index().drop(columns="index")
```

#### Observations: Changes Between 2006 and 2025

China recorded the largest widening of its tyre-trade deficit, increasing by AUD 936.008 million. Japan and the United States recorded the next-largest increases in deficits.

Canada, Singapore, Hong Kong, Germany, the United Kingdom and the Republic of Korea recorded smaller deficits in 2025 than in 2006, but remained in deficit.

Both New Zealand and the United Arab Emirates moved from deficits in 2006 to surpluses in 2025. New Zealand recorded the largest positive change in trade balance, at AUD 47.455 million.

Overall, 14 of the 22 matched partners had more negative trade balances, while eight had more positive balances.

This comparison measures changes between two endpoints. It does not describe fluctuations during the intervening years or explain their causes.
